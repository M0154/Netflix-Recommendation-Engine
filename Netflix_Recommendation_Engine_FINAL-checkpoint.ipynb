## Step 2: Load the Dataset

> **Note:** Place `combined_data_1.txt` and `movie_titles.csv` in the same folder as this notebook.


```python
# ── 2a: Load movie titles ──
movies = pd.read_csv(
    'movie_titles.csv',
    header=None,
    names=['MovieID', 'Year', 'Title'],
    encoding='latin-1',
    on_bad_lines='skip'
)
print(f'Movies loaded: {movies.shape[0]:,} titles')
movies.head()
```

```python
# ── 2b: Assign Genre via keyword classification ──
# The Netflix Prize dataset has no genre column — we infer it from the movie title.

genre_keywords = {
    'Action':      ['action','war','combat','fighter','battle','commando','soldier','warrior','spy','gun','mission','strike','raid','sniper','assault'],
    'Comedy':      ['comedy','funny','laugh','humor','sitcom','fun','joke','gag','prank','spoof','parody','slapstick'],
    'Drama':       ['drama','life','story','tale','journey','family','heart','soul','tears','tragedy','grief','loss','struggle','redemption'],
    'Horror':      ['horror','scary','fear','ghost','monster','demon','evil','haunted','curse','blood','zombie','vampire','nightmare','terror','scream','killer'],
    'Romance':     ['romance','love','romantic','wedding','bride','kiss','affair','passion','desire'],
    'Sci-Fi':      ['sci-fi','science fiction','space','alien','robot','future','galaxy','planet','star','cosmic','quantum','cyber','clone'],
    'Thriller':    ['thriller','suspense','mystery','crime','detective','murder','chase','heist','prison','escape','hitman','assassin','conspiracy'],
    'Documentary': ['documentary','true story','history','historical','biography','biopic','nature','wildlife','national geographic'],
    'Animation':   ['animation','animated','cartoon','anime','pixar','disney','dreamworks','children','kids','fairy'],
    'Fantasy':     ['fantasy','magic','wizard','witch','dragon','elf','mythical','legend','epic','sword','knight','kingdom','medieval','supernatural'],
}

def classify_genre(title):
    if not isinstance(title, str):
        return 'Drama'
    t = title.lower()
    for genre, kws in genre_keywords.items():
        if any(kw in t for kw in kws):
            return genre
    return 'Drama'  # default for unclassified titles

movies['Genre'] = movies['Title'].apply(classify_genre)
print('Genre Distribution in Movie Catalogue:')
print(movies['Genre'].value_counts())
```

```python
# ── 2c: Parse ratings from combined_data_1.txt ──
print('Parsing ratings file ... (takes ~30 sec for 600k rows)')

data = []
movie_id = None
LIMIT = 600_000   # Increase this for more data

with open('combined_data_1.txt', 'r') as f:
    for line in f:
        line = line.strip()
        if line.endswith(':'):
            movie_id = int(line[:-1])
        else:
            parts = line.split(',')
            if len(parts) == 3:
                customer_id, rating, date = parts
                data.append([movie_id, int(customer_id), int(rating), date])
                if len(data) >= LIMIT:
                    break

df = pd.DataFrame(data, columns=['MovieID', 'CustomerID', 'Rating', 'Date'])
df['Date'] = pd.to_datetime(df['Date'])

# Merge with movie metadata
df = df.merge(movies[['MovieID', 'Title', 'Genre', 'Year']], on='MovieID', how='left')

print(f'\n✅ Dataset ready!')
print(f'   Shape            : {df.shape}')
print(f'   Unique Customers : {df["CustomerID"].nunique():,}')
print(f'   Unique Movies    : {df["MovieID"].nunique():,}')
print(f'   Date Range       : {df["Date"].min().date()} → {df["Date"].max().date()}')
print(f'   Avg Rating       : {df["Rating"].mean():.3f}')
df.head()
```

## Step 5: Objective 2 — Recommendation Model

**Algorithm: User-Based Collaborative Filtering with Cosine Similarity**

Steps:
1. Build a User–Movie rating matrix from active users
2. Compute cosine similarity between every pair of users
3. For a target user, find the most similar users
4. From movies those similar users rated highly (that the target hasn't seen), pick the **best movie per genre**


```python
# Build User–Movie matrix from top active users
top_users = (
    df.groupby('CustomerID')
      .agg(n_ratings=('Rating','count'), n_genres=('Genre','nunique'))
      .reset_index()
      .query('n_ratings >= 5')
      .sort_values('n_ratings', ascending=False)
      .head(500)['CustomerID'].tolist()
)

sample_df = df[df['CustomerID'].isin(top_users)]

user_movie_matrix = sample_df.pivot_table(
    index='CustomerID',
    columns='Title',
    values='Rating',
    aggfunc='mean'
).fillna(0)

print(f'User-Movie Matrix shape: {user_movie_matrix.shape}')
print(f'  Rows = {user_movie_matrix.shape[0]} users')
print(f'  Cols = {user_movie_matrix.shape[1]} movies')
user_movie_matrix.head(3)
```

```python
# Compute Cosine Similarity
user_similarity = cosine_similarity(user_movie_matrix)
user_sim_df = pd.DataFrame(
    user_similarity,
    index=user_movie_matrix.index,
    columns=user_movie_matrix.index
)

# Heatmap of first 15 users
plt.figure(figsize=(12, 7))
sns.heatmap(user_sim_df.iloc[:15, :15], cmap='YlOrRd',
            linewidths=0.3, annot=True, fmt='.2f', annot_kws={'size':7},
            cbar_kws={'label': 'Cosine Similarity'})
plt.title('User–User Cosine Similarity Matrix (Top 15 Users)', fontsize=13, fontweight='bold')
plt.xlabel('Customer ID')
plt.ylabel('Customer ID')
plt.xticks(rotation=45, ha='right', fontsize=7)
plt.yticks(fontsize=7)
plt.tight_layout()
plt.show()
```

```python
# Recommendation functions

def get_similar_users(user_id, n=20):
    """Returns top-N most similar users to user_id."""
    if user_id not in user_sim_df.index:
        return []
    return (user_sim_df[user_id]
            .drop(user_id)
            .sort_values(ascending=False)
            .head(n).index.tolist())


def recommend_movies_per_genre(user_id, n_similar=20, min_ratings=3):
    """
    Collaborative Filtering Recommender.
    For a given user, finds similar users and returns
    the best unseen movie in each genre.
    """
    watched = set(df[df['CustomerID'] == user_id]['Title'].unique())
    similar_users = get_similar_users(user_id, n=n_similar)

    if similar_users:
        pool = df[
            df['CustomerID'].isin(similar_users) &
            ~df['Title'].isin(watched)
        ]
    else:
        pool = df[~df['Title'].isin(watched)]  # fallback: all users

    if pool.empty:
        return recommend_global_top()

    genre_recs = (
        pool.groupby(['Genre', 'Title'])['Rating']
            .agg(Avg_Rating='mean', Num_Ratings='count')
            .reset_index()
    )
    genre_recs = genre_recs[genre_recs['Num_Ratings'] >= min_ratings]

    best = (
        genre_recs.sort_values('Avg_Rating', ascending=False)
                  .groupby('Genre').first()
                  .reset_index()
                  [['Genre', 'Title', 'Avg_Rating', 'Num_Ratings']]
    )
    best['Avg_Rating'] = best['Avg_Rating'].round(3)
    return best


def recommend_global_top(min_ratings=10):
    """Fallback: global highest-rated movie per genre."""
    stats = (
        df.groupby(['Genre', 'Title'])['Rating']
          .agg(Avg_Rating='mean', Num_Ratings='count')
          .reset_index()
    )
    stats = stats[stats['Num_Ratings'] >= min_ratings]
    best = (
        stats.sort_values('Avg_Rating', ascending=False)
             .groupby('Genre').first()
             .reset_index()
             [['Genre', 'Title', 'Avg_Rating', 'Num_Ratings']]
    )
    best['Avg_Rating'] = best['Avg_Rating'].round(3)
    return best


print('✅ Recommendation model ready!')
```

```python
# Run for most active user
target_user = top_users[0]
print(f'🎯 Target User: {target_user}')

history = df[df['CustomerID'] == target_user][['Genre','Title','Rating']]
print(f'   Total ratings by this user: {len(history)}')
print(f'   Genres watched: {history["Genre"].nunique()}')
print('\n  Sample Watch History:')
print(history.head(8).to_string(index=False))

recs = recommend_movies_per_genre(target_user)
print(f'\n🎬 Recommendations for User {target_user}:')
recs
```

```python
# Global Best Movie per Genre
global_top = recommend_global_top()
print('=== Global Best Suited Movie per Genre (Objective 2 Answer) ===')
global_top
```

## Step 6: Objective 3 — Best & Worst Rated Genres


```python
genre_rating = df.groupby('Genre')['Rating'].agg(
    Mean_Rating   = 'mean',
    Median_Rating = 'median',
    Std_Dev       = 'std',
    Total_Ratings = 'count'
).round(3).reset_index().sort_values('Mean_Rating', ascending=False)

print('=== Genre Rating Statistics (from real Netflix data) ===')
print(genre_rating.to_string(index=False))

best_genre  = genre_rating.iloc[0]
worst_genre = genre_rating.iloc[-1]
print(f'\n🏆 Best Rated Genre  : {best_genre["Genre"]:12s} — avg ⭐ {best_genre["Mean_Rating"]}')
print(f'📉 Worst Rated Genre : {worst_genre["Genre"]:12s} — avg ⭐ {worst_genre["Mean_Rating"]}')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sorted_g = genre_rating.sort_values('Mean_Rating')
bar_cols = ['#2ecc71' if g == best_genre['Genre']
            else '#e74c3c' if g == worst_genre['Genre']
            else '#3498db' for g in sorted_g['Genre']]

# Bar chart
axes[0].barh(sorted_g['Genre'], sorted_g['Mean_Rating'], color=bar_cols, edgecolor='white')
axes[0].axvline(df['Rating'].mean(), color='orange', linestyle='--', lw=2,
                label=f'Overall avg ({df["Rating"].mean():.3f})')
axes[0].set_title('⭐ Average Rating by Genre\n🟢 Best  |  🔴 Worst  |  🔵 Others',
                  fontsize=13, fontweight='bold')
axes[0].set_xlabel('Average Star Rating')
axes[0].set_xlim(0, 5.3)
axes[0].legend(fontsize=9)
for i, (g, v) in enumerate(zip(sorted_g['Genre'], sorted_g['Mean_Rating'])):
    axes[0].text(v + 0.04, i, f'{v:.3f}', va='center', fontsize=9)

# Box plot
genre_order = genre_rating.sort_values('Mean_Rating', ascending=False)['Genre'].tolist()
genre_data  = [df[df['Genre'] == g]['Rating'].values for g in genre_order]
bp = axes[1].boxplot(genre_data, vert=False, patch_artist=True,
                     labels=genre_order, notch=False)
box_colors = sns.color_palette('Set2', len(genre_order))
for patch, c in zip(bp['boxes'], box_colors):
    patch.set_facecolor(c)
    patch.set_alpha(0.8)
axes[1].set_title('📊 Rating Distribution per Genre\n(Box Plot)', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Star Rating')

plt.suptitle('Genre Ratings Analysis — Best vs Worst', fontsize=15, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Violin Plot
plt.figure(figsize=(14, 6))
genre_order_v = genre_rating.sort_values('Mean_Rating', ascending=False)['Genre'].tolist()
sns.violinplot(data=df, x='Genre', y='Rating', order=genre_order_v,
               palette='Set2', inner='quartile', cut=0)
plt.axhline(df['Rating'].mean(), color=NETFLIX_RED, linestyle='--', lw=2,
            label=f'Overall mean ({df["Rating"].mean():.3f})')
plt.title('Rating Distribution by Genre (Violin Plot)', fontsize=14, fontweight='bold')
plt.xlabel('Genre')
plt.ylabel('Star Rating')
plt.xticks(rotation=20)
plt.legend()
plt.tight_layout()
plt.show()
```
